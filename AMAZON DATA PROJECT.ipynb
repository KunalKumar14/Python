### Changing Data Types of columns From object to Float

```python
# Changing the data type of discounted price and actual price 
Amazon['discounted_price'] = Amazon['discounted_price'].str.replace("₹",'',regex=False) .str.replace(",", "", regex=False).astype('Float64') 
```

```python
Amazon['actual_price'] = Amazon['actual_price'].str.replace("₹",'',regex=False).str.replace(",", "", regex=False).astype('Float64')
```

```python
Amazon['discount_percentage'] = Amazon['discount_percentage'].str.replace('%','').astype('float64')
```

```python
Amazon['discount_percentage'] = Amazon['discount_percentage']/ 100
```

```python
Amazon['rating'].value_counts()
```

```python
# check the |        1 row 
Amazon.query('rating == " | "')
```

```python
# changing Rating Columns Data type 
Amazon['rating'] = Amazon['rating'].str.replace('|', '3.9').astype('float64')
```

```python
# changing 'rating_count'  coloumn Data type

Amazon['rating_count'] = Amazon['rating_count'].str.replace(',','').astype('Float64')
```

### Adding New Columns

```python
# Absolute savings from the discount
Amazon['savings_amount'] = Amazon['actual_price'] - Amazon['discounted_price']
```

```python

# Rating category
Amazon['rating_category'] = pd.cut(
    Amazon['rating'],
    bins=[0, 3, 4, 4.5, 5],
    labels=['Low (<=3)', 'Average (3-4)', 'Good (4-4.5)', 'Excellent (4.5-5)']
)
```

```python
Amazon['main_category'] = Amazon['category'].str.split('|').str[0]
```

### Aggergate

```python
amazon_category_summary = (
    Amazon.groupby('main_category')
    .agg(
        category_avg_rating=('rating', 'mean'),
        category_avg_price=('discounted_price', 'mean'),
        category_avg_discount=('discount_percentage', 'mean'),
        category_total_reviews=('rating_count', 'sum'),
        category_product_count=('product_id', 'count'),
    )
    .reset_index()
)
amazon_category_summary = amazon_category_summary.sort_values('category_product_count', ascending=False)
```

### Pricing & Discounts

```python
plt.figure(figsize=(10, 6))
sns.histplot(Amazon['discount_percentage'], bins=20, color='orange', kde=True)
plt.title('Distribution of Discount Percentage')
plt.xlabel('Discount (%)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=Amazon, x='actual_price', y='discounted_price',
                 hue='discount_percentage', palette='viridis', alpha=0.6)
plt.title('Actual Price vs Discounted Price')
plt.xlabel('Actual Price (₹)')
plt.ylabel('Discounted Price (₹)')
plt.xscale('log')
plt.yscale('log')
plt.tight_layout()
plt.show()
```

```python
amazon_avg_discount_by_cat = ( Amazon.groupby('main_category')['discount_percentage'] .mean().sort_values(ascending=False).head(10))

plt.figure(figsize=(10, 6))
sns.barplot(x=amazon_avg_discount_by_cat.values, y=amazon_avg_discount_by_cat.index, palette='magma')
plt.title('Top 10 Categories by Average Discount (%)')
plt.xlabel('Average Discount (%)')
plt.ylabel('Category')
plt.tight_layout()
plt.show()
```

### Insight Summary

| # | Insight | Business Implication |
|---|---------|----------------------|
| 1 | A handful of categories dominate the catalog | Category strategy should focus depth on top categories rather than breadth |
| 2 | Heavy discounting is the norm, not the exception | Discount-driven pricing appears to be a core conversion lever on this platform |
| 3 | Ratings skew positive and cluster tightly | A rating alone is a weak differentiator; review volume matters more for trust |
| 4 | Price is a poor predictor of rating | Premium pricing doesn't guarantee better perceived quality — value positioning matters |
| 5 | Review engagement is very unevenly distributed | A small set of "hero" products drive most social proof; new listings need a push strategy |
| 6 | High-value (rating-per-rupee) products exist across price points | A "best value" filter could be a useful product-discovery feature |

## 9. Summary & Conclusions

**Data cleaning performed:**
- Checked and handled data types, duplicate rows/product IDs, missing values,
  and text inconsistencies (whitespace, invalid entries).
- Converted price, discount, rating, and rating-count columns from text to
  proper numeric types.
- Engineered new columns (`main_category`, `savings_amount`, `price_bucket`,
  `discount_bucket`, `rating_category`, `value_score`).
- Combined a category-level summary table back into the main product table via
  `pd.merge`.

**Data transformation performed:**
- Log-transformed skewed price and review-count features.
- Applied Min-Max normalization and Z-score standardization to key numeric
  features.
- Label-encoded and one-hot-encoded categorical variables.
- Built a pivot table of rating category vs. product category.

**Key findings:**
- The dataset is dominated by electronics accessories (cables, chargers, etc.).
- Discounts are common and often steep, with many products discounted well
  above 50%.
- Ratings cluster tightly between roughly 3.5 and 4.5 - most listed products
  are reasonably well received.
- Price and rating show weak correlation - expensive items aren't necessarily
  rated higher.
- A small number of categories account for the bulk of both listings and
  reviews.
